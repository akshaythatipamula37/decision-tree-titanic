```python
df = df.drop('Cabin',axis = 1)
df.drop('PassengerId',axis=1, inplace=True)
df.drop('Name',axis =1,inplace = True)
df.drop('Ticket',axis =1,inplace = True)
```

```python
df.columns
```

```python
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df_encoded = pd.get_dummies(df, columns=["Sex", "Embarked"])
df_encoded.head()
```

```python
x = df_encoded[['Pclass','Age','SibSp','Parch','Sex_male','Sex_female','Embarked_C','Embarked_Q','Embarked_S']]
y = df_encoded['Survived']
```

```python
import sklearn.tree
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(max_depth=3, random_state=42)
```

```python
model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(x, y)
```

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(20,10))
plot_tree(model, feature_names=x.columns, class_names=["Died", "Survived"], filled=True)
plt.show()
```
