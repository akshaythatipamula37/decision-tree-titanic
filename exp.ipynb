```python
import pandas as pd
import numpy as np

# Check the training data
print("Training samples:", len(x_train))
print("Features:", list(x_train.columns))

# Count the classes at the root
print("\nTarget class counts:")
print(y_train.value_counts().sort_index())

# Calculate the root Gini impurity
counts = y_train.value_counts().to_numpy()
probabilities = counts / counts.sum()

root_gini = 1 - np.sum(probabilities ** 2)

print("\nClass probabilities:", probabilities)
print("Root Gini:", root_gini)
```

```python
def gini_impurity(y):
    probabilities = y.value_counts(normalize=True)
    return 1 - np.sum(probabilities ** 2)


def evaluate_split(X, y, feature, threshold):
    left_mask = X[feature] <= threshold

    y_left = y[left_mask]
    y_right = y[~left_mask]

    if len(y_left) == 0 or len(y_right) == 0:
        return None

    gini_left = gini_impurity(y_left)
    gini_right = gini_impurity(y_right)

    weighted_gini = (
        len(y_left) / len(y) * gini_left
        + len(y_right) / len(y) * gini_right
    )
    print("feature :",feature,"\n",
                "threshold :",threshold, "\n",
                "left_samples :",len(y_left),"\n"
                "right_samples :",len(y_right),"\n"
                "left_gini :",gini_left, "\n",
                "right_gini :", gini_right,"\n",
                "weighted_gini :", weighted_gini,"\n",
                "gini_reduction :", gini_impurity(y) - weighted_gini, "\n"*2)
    # return {
    #     "feature": feature,
    #     "threshold": threshold,
    #     "left_samples": len(y_left),
    #     "right_samples": len(y_right),
    #     "left_gini": gini_left,
    #     "right_gini": gini_right,
    #     "weighted_gini": weighted_gini,
    #     "gini_reduction": gini_impurity(y) - weighted_gini
    # }
    

for i in list(x_train.columns):
    evaluate_split(x_train, y_train, i, 0.5)
```
